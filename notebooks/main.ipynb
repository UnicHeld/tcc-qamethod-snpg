## Embedder

```python
import os
import json
from backend.app.services.embedder import Embedder
from backend.app.services.qdrant_manager import QdrantManager
```

```python
JSON_PATH = "/home/thehprogrammer/Projects/python/qa-method/data/json/"
openai_api_key = os.environ.get("OPENAI_API_KEY")
qdrant_api_key = os.environ.get("QDRANT_API_KEY")
```

```python
if not openai_api_key:
    raise ValueError("OPENAI_API_KEY não está definida nas variáveis de ambiente.")

if not qdrant_api_key:
    raise ValueError("QDRANT_API_KEY não está definida nas variáveis de ambiente.")
```

```python
embedder = Embedder(api_key=openai_api_key)
```

```python
updated_json_content = embedder.add_embedding_to_json(JSON_PATH)

if updated_json_content:
    print(json.dumps(updated_json_content, ensure_ascii=False, indent=4))
else:
    print("Erro ao adicionar embedding ao JSON.")
```
